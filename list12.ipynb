```python
# List is Used To Store Multivalue in Variable
#(1).List Operator
a = [10,20,30]
b = [40,50]
c = [10,20,30]
#Concatenation Operator
print("Addition:",a+b)
#Repitition Operator
print("Repetition:",a*2)
#Membership Operator
print("Membership:",20 in a)
#Not Membership Operator
print("Not Membership:",20 not in a)
#Identity Operator
print("Identity:",a is c)
#Not Identity Operator
print("Not Identity:",a is not c)
#Equality Operator
print("Equality:",a == c)
#Inequality Operator
print("Inequality:",a != c)
#(2).List Operation
num = [1,2,3,4,5]
#Acess an Element
print("Acess:",num[1])
#Slicing an Element
print("Slicing:",num[2:4])
#Update an Element
num [1] = 2
print("Update:",num)
#Delete an Element
del num[1]
print("Delete:",num)
#Append an Element
num.append(6)
print("Append:",num)
#Insert an Element
num.insert(2,3)
print("Insert:",num)
#Remove an Element
num.remove(3)
print("Remove:",num)
#Extend an Element
num.extend([7,8])
print("Extend:",num)
#Pop an Element
num.pop()
print("Pop:",num)
#Reverse an Element
num.reverse()
print("Reverse:",num)
#Sort an Element
num.sort()
print("Sort:",num)
#Count an Element
print("Count:",num.count(2))
#Copy an Element
num1 = num.copy()
print("Copy:",num1)
#Clear an Element
num.clear()
print("Clear:",num)
#(3).Built-in Function
lab = [10,20,30,40,50,60]
#Length of List
print("Length:",len(lab))
#Maximum of List
print("Maximum:",max(lab))
#Minimum of List
print("Minimum:",min(lab))
#Sum of List
print("Sum:",sum(lab))

```
